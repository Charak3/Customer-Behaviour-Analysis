```python
import pandas as pd
```

```python
df=pd.read_csv("customer_shopping_behavior.csv")
```

```python
#replacing the null values in review rating columns with the median review rating as per the category 

df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns =df.columns.str.lower().str.replace(' ','_')
```

```python
df.columns
```

```python
df= df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

```python
#creating a column age_group

labels =['Young Adult','Adult ','Middle Aged ','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
df[['age_group']].head(10)
```

```python
# Making the numerical colmn fot frequency of purchases column

df['frequency_of_purchases'].unique()
```

```python
frequency_mapping ={'Fortnightly':14, 'Weekly':7, 'Annually':365, 'Quarterly':90, 'Bi-Weekly':14,
       'Monthly':30, 'Every 3 Months':90}

df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
#Which category is loved by customers the most
category_counts = df['category'].value_counts()

plt.bar(category_counts.index, category_counts.values)

plt.title('Total Number of Purchases by Category', fontsize=13)
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Number of Items Sold', fontsize=12)

plt.show()
```

```python
### Connecting python script to mysql

from sqlalchemy import create_engine
import pandas as pd

# Step 1: Connect to MySQL
username = "root"          # Default MySQL username is usually 'root'
password = "qwerty12"      # Your MySQL password
host = "localhost"         # If running locally
port = "3306"              # Default MySQL port is 3306 (changed from 5432)
database = "customer_behaviour"

# The engine string changes to specify 'mysql' and the 'pymysql' driver
engine = create_engine(f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}")

# Step 2: Load DataFrame into MySQL
table_name = "customer"

# Write the data. This part remains exactly the same!
df.to_sql(table_name, con=engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
